```python
def format_chatml(messages):
    """Render a conversation as ChatML text."""
    out = []
    for m in messages:
        out.append(f"<|im_start|>{m['role']}\n{m['content']}<|im_end|>\n")
    return "".join(out)


# 2-message case
print(format_chatml(train[0]["messages"]))
```

```python
from transformers import AutoTokenizer

tok = AutoTokenizer.from_pretrained(MODEL_ID)

print("special:", tok.special_tokens_map)
print("vocab len:", len(tok))

s = "<|im_start|>assistant\nhi<|im_end|>\n"
ids = tok(s, add_special_tokens=False).input_ids
print(ids)
print(tok.convert_ids_to_tokens(ids))
```

```python
IGNORE = -100

def encode(messages, tok, max_seq=MAX_SEQ):
    input_ids, labels = [], []

    for m in messages:
        header = f"<|im_start|>{m['role']}\n"
        body = f"{m['content']}<|im_end|>\n"

        h_ids = tok(header, add_special_tokens=False).input_ids
        b_ids = tok(body, add_special_tokens=False).input_ids

        input_ids += h_ids + b_ids
        labels += [IGNORE] * len(h_ids)
        labels += b_ids if m["role"] == "assistant" else [IGNORE] * len(b_ids)

    return {
        "input_ids": input_ids[:max_seq],
        "labels": labels[:max_seq],
    }
```

```python
ex = encode(multi["messages"], tok)

print(f"length: {len(ex['input_ids'])}")
print(f"trained tokens: {sum(1 for l in ex['labels'] if l != IGNORE)}")
print()

for i, l in zip(ex["input_ids"], ex["labels"]):
    mark = "  " if l == IGNORE else "->"
    print(f"{mark} {tok.convert_ids_to_tokens([i])[0]!r}")
```

```python
bad = 0
for r in train:
    ex = encode(r["messages"], tok)
    if all(l == IGNORE for l in ex["labels"]):
        bad += 1
print(f"no trained tokens at max_seq={MAX_SEQ}: {bad}")
```

```python
def build(rows, tok, max_seq=MAX_SEQ):
    kept, dropped = [], 0
    for r in rows:
        ex = encode(r["messages"], tok, max_seq=max_seq)
        if all(l == IGNORE for l in ex["labels"]):
            dropped += 1
            continue
        kept.append(ex)
    print(f"kept {len(kept)}, dropped {dropped}")
    return kept

train_enc = build(train, tok)
test_enc = build(test, tok)
```

```python
tot = sum(len(e["input_ids"]) for e in train_enc)
trn = sum(1 for e in train_enc for l in e["labels"] if l != IGNORE)
print(f"{len(train_enc)} examples, {tot:,} tokens, {trn:,} trained ({100*trn/tot:.1f}%)")
```

```python
PAD_ID = tok.pad_token_id  # 151643

def collate(batch):
    """
    Pad to the longest example IN THIS BATCH, not to MAX_SEQ.

    p50 is 232 tokens but max is 1024. Padding every batch to 1024 would
    waste ~4x the compute on nothing. Dynamic padding costs one line.
    """
    n = max(len(b["input_ids"]) for b in batch)

    input_ids, labels, attn = [], [], []
    for b in batch:
        pad = n - len(b["input_ids"])
        input_ids.append(b["input_ids"] + [PAD_ID] * pad)
        labels.append(b["labels"] + [IGNORE] * pad)      # padding never contributes loss
        attn.append([1] * len(b["input_ids"]) + [0] * pad)

    return {
        "input_ids": torch.tensor(input_ids, dtype=torch.long),
        "labels": torch.tensor(labels, dtype=torch.long),
        "attention_mask": torch.tensor(attn, dtype=torch.long),
    }
```

```python
from torch.utils.data import DataLoader

BATCH = 4

loader = DataLoader(
    train_enc,
    batch_size=BATCH,
    shuffle=True,
    collate_fn=collate,
    drop_last=True,
)

batch = next(iter(loader))

for k, v in batch.items():
    print(f"{k:16} {tuple(v.shape)}  {v.dtype}")

print()
print("padding fraction:", 1 - batch["attention_mask"].float().mean().item())
print("trained tokens:  ", (batch["labels"] != IGNORE).sum().item())
print("total positions: ", batch["labels"].numel())
```

## LORA

```python
import torch.nn as nn

class LoRALinear(nn.Module):
    """
    Wraps a frozen nn.Linear with a trainable low-rank update.

        out = base(x) + (alpha / r) * B(A(x))

    B is zero-initialized, so the adapter is an exact no-op until the first
    optimizer step.

    Adapters stay in fp32 while the frozen base is bf16. The base holds the
    memory; the adapters hold the precision that matters. We cast at the
    boundary rather than casting the adapters down.
    """

    def __init__(self, base: nn.Linear, r: int = 32, alpha: int = 64, dropout: float = 0.0):
        super().__init__()
        self.base = base
        for p in self.base.parameters():
            p.requires_grad = False

        self.r = r
        self.scaling = alpha / r

        self.lora_A = nn.Linear(base.in_features, r, bias=False)
        self.lora_B = nn.Linear(r, base.out_features, bias=False)
        self.dropout = nn.Dropout(dropout) if dropout > 0 else nn.Identity()

        nn.init.kaiming_uniform_(self.lora_A.weight, a=5 ** 0.5)
        nn.init.zeros_(self.lora_B.weight)

    def forward(self, x):
        base_out = self.base(x)
        lora_in = self.dropout(x).to(self.lora_A.weight.dtype)
        lora_out = self.lora_B(self.lora_A(lora_in))
        return base_out + self.scaling * lora_out.to(base_out.dtype)
```

```python
torch.manual_seed(0)

base = nn.Linear(1024, 2048, bias=False)   # q_proj shape
x = torch.randn(2, 7, 1024)

wrapped = LoRALinear(base, r=32, alpha=64)

with torch.no_grad():
    diff = (wrapped(x) - base(x)).abs().max().item()

print(f"max deviation from base: {diff}")
assert diff == 0.0, "LoRA is not a no-op at init"

# now perturb B and confirm the path is actually live
with torch.no_grad():
    wrapped.lora_B.weight.normal_(0, 0.01)
    diff2 = (wrapped(x) - base(x)).abs().max().item()

print(f"after perturbing B: {diff2}")
assert diff2 > 0, "LoRA path is dead"

print("\nshapes:")
print(" A:", tuple(wrapped.lora_A.weight.shape))
print(" B:", tuple(wrapped.lora_B.weight.shape))
print(" trainable:", sum(p.numel() for p in wrapped.parameters() if p.requires_grad))
```

```python
TARGETS = ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"]

def apply_lora(model, targets=TARGETS, r=32, alpha=64, dropout=0.0):
    """Walk the module tree and swap matching nn.Linear layers for LoRALinear."""
    replaced = 0
    for name, module in model.named_modules():
        for child_name, child in list(module.named_children()):
            if child_name in targets and isinstance(child, nn.Linear):
                setattr(module, child_name, LoRALinear(child, r, alpha, dropout))
                replaced += 1
    return replaced
```

```python
from transformers import AutoModelForCausalLM

try:
    del model
except NameError:
    pass
torch.cuda.empty_cache()

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    dtype=torch.bfloat16,
    attn_implementation="sdpa",
)

for p in model.parameters():
    p.requires_grad = False

n = apply_lora(model, r=32, alpha=64)
model = model.cuda()

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"replaced {n} layers")
print(f"trainable {trainable/1e6:.2f}M / {total/1e6:.1f}M ({100*trainable/total:.2f}%)")
print(f"vram {torch.cuda.memory_allocated()/1e9:.2f} GB")

# confirm the dtype split is what we intended
p = model.model.layers[0].self_attn.q_proj
print("base dtype:   ", p.base.weight.dtype)
print("adapter dtype:", p.lora_A.weight.dtype)
```

```python
out.loss.backward()

with_grad = [n for n, p in model.named_parameters()
             if p.requires_grad and p.grad is not None]
print(f"params with grads: {len(with_grad)} (expect 392)")

# lora_A's gradient is exactly zero on the first backward, because dL/dA
# flows through B, which is still zero. Expected, not a bug.
a = model.model.layers[0].self_attn.q_proj.lora_A.weight.grad
b = model.model.layers[0].self_attn.q_proj.lora_B.weight.grad
print(f"layer0 q_proj  |grad A| = {a.abs().max():.3e}   |grad B| = {b.abs().max():.3e}")

norm = torch.nn.utils.clip_grad_norm_(
    [p for p in model.parameters() if p.requires_grad], 1.0
)
print("grad norm:", norm.item())
print(f"vram after backward: {torch.cuda.memory_allocated()/1e9:.2f} GB")
```

```python
from torch.optim import AdamW

STEPS = 100
GRAD_ACCUM = 4
LR = 2e-4
WARMUP = 10

params = [p for p in model.parameters() if p.requires_grad]
opt = AdamW(params, lr=LR, weight_decay=0.01, betas=(0.9, 0.999))

sched = torch.optim.lr_scheduler.OneCycleLR(
    opt, max_lr=LR, total_steps=STEPS,
    pct_start=WARMUP / STEPS, anneal_strategy="cos",
)

print(f"optimizing {sum(p.numel() for p in params)/1e6:.2f}M params")
print(f"effective batch = {BATCH} x {GRAD_ACCUM} = {BATCH * GRAD_ACCUM}")
print(f"{STEPS} steps covers {STEPS * BATCH * GRAD_ACCUM} examples "
      f"({STEPS * BATCH * GRAD_ACCUM / len(train_enc):.2f} epochs)")
```

```python
PROBES = [
    "Write a two-sentence product description for noise-cancelling headphones.",
    "Explain what a database index does, in one short paragraph.",
]

IM_END = tok.convert_tokens_to_ids("<|im_end|>")

@torch.no_grad()
def chat(prompt, max_new_tokens=120):
    model.eval()
    text = f"<|im_start|>user\n{prompt}<|im_end|>\n<|im_start|>assistant\n"
    ids = tok(text, add_special_tokens=False, return_tensors="pt").input_ids.cuda()
    out = model.generate(
        ids,
        max_new_tokens=max_new_tokens,
        do_sample=False,
        eos_token_id=IM_END,
        pad_token_id=tok.pad_token_id,
    )
    model.train()
    return tok.decode(out[0][ids.shape[-1]:], skip_special_tokens=False)

before = [chat(p) for p in PROBES]
for p, b in zip(PROBES, before):
    print(f"### {p}\n{b}\n")
```

```python
import time
from itertools import islice

def endless(loader):
    while True:
        yield from loader

stream = endless(loader)
model.train()
opt.zero_grad(set_to_none=True)

log = []
t0 = time.time()

for step in range(1, STEPS + 1):
    total = 0.0
    for _ in range(GRAD_ACCUM):
        batch = {k: v.cuda(non_blocking=True) for k, v in next(stream).items()}
        loss = model(**batch).loss
        # Scale so the accumulated gradient equals the mean over the full
        # effective batch, not the sum of four separate means.
        (loss / GRAD_ACCUM).backward()
        total += loss.item() / GRAD_ACCUM

    gnorm = torch.nn.utils.clip_grad_norm_(params, 1.0)
    opt.step()
    sched.step()
    opt.zero_grad(set_to_none=True)

    log.append({"step": step, "loss": total, "gnorm": gnorm.item(), "lr": sched.get_last_lr()[0]})

    if step % 5 == 0 or step == 1:
        print(f"step {step:4d}  loss {total:.4f}  gnorm {gnorm:5.2f}  "
              f"lr {sched.get_last_lr()[0]:.2e}  {time.time()-t0:5.0f}s")

print(f"\ndone in {time.time()-t0:.0f}s")
print(f"peak vram {torch.cuda.max_memory_allocated()/1e9:.2f} GB")
```

```python
import torch

model.eval()
ranks, probs_ = [], []

with torch.no_grad():
    for ex in test_enc[:32]:
        ids = torch.tensor([ex["input_ids"]]).cuda()
        lab = torch.tensor(ex["labels"])
        # positions where the NEXT token is <|im_end|>
        pos = [i - 1 for i in range(1, len(ex["input_ids"]))
               if ex["input_ids"][i] == IM_END and lab[i] != IGNORE]
        if not pos:
            continue
        logits = model(ids).logits[0]
        for p in pos:
            d = logits[p].softmax(-1)
            probs_.append(d[IM_END].item())
            ranks.append((d > d[IM_END]).sum().item())

model.train()
print(f"n={len(ranks)}  median rank={sorted(ranks)[len(ranks)//2]}  "
      f"mean P={sum(probs_)/len(probs_):.4f}")
```

```python
def load_lora(ckpt_dir, device="cuda"):
    ckpt_dir = Path(ckpt_dir)
    meta = json.loads((ckpt_dir / "adapter_config.json").read_text())

    m = AutoModelForCausalLM.from_pretrained(
        meta["base_model"], dtype=torch.bfloat16, attn_implementation="sdpa",
    )
    for p in m.parameters():
        p.requires_grad = False

    n = apply_lora(m, targets=meta["targets"], r=meta["r"],
                   alpha=meta["alpha"], dropout=meta["dropout"])

    sd = torch.load(ckpt_dir / "adapter.pt", map_location="cpu")
    missing, unexpected = m.load_state_dict(sd, strict=False)

    # strict=False will happily load NOTHING and not complain. Check explicitly.
    lora_missing = [k for k in missing if "lora_" in k]
    assert not lora_missing, f"adapters not loaded: {lora_missing[:3]}"
    assert not unexpected, f"unexpected keys: {unexpected[:3]}"

    print(f"replaced {n}, loaded {len(sd)} adapter tensors")
    return m.to(device), meta
```
